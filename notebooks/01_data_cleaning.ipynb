```python
df = df.drop_duplicates()
```

**Step 6 — Check Data Types**

```python
df.dtypes
```

**Step 7 — Convert Month Name to Month Number**

```python
month_map = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

df["Month"] = df["Month"].map(month_map)
```

```python
df.head()
```


**Step 8 — Create Date Column**

```python
df["Date"] = pd.to_datetime(
    dict(
        year=df["Year"],
        month=df["Month"],
        day=1
    )
)
```

**Step 9 — Rearrange Columns**

```python
df = df[
    [
        "Date",
        "Year",
        "Month",
        "State",
        "Vehicle_Category",
        "EV_Sales"
    ]
]
```

**Step 10 — Sort the Data**

```python
df = df.sort_values(
    ["Date", "State", "Vehicle_Category"]
)
```

```python
df
```

**Step 11 — Reset the Index**

```python
df.reset_index(drop=True, inplace=True)
```

**Step 12 — Save the Cleaned Dataset**
Create the output folder if needed:

```python
os.makedirs("../data/processed", exist_ok=True)
```

```python
df.to_csv(
    "../data/processed/ev_sales_cleaned.csv",
    index=False
)
```
