```python
import pandas as pd
```

```python
df = pd.read_csv(
    "../data/processed/ev_sales_feature_engineered.csv",
    parse_dates=["Date"]
)

df.head()
```

```python
df.columns
```

```python
df = pd.get_dummies(
    df,
    columns=["State", "Vehicle_Category"],
    drop_first=True
)
```

```python
df.head()
```

```python
y = df["EV_Sales"]
```

```python
X = df.drop(
    columns=[
        "Date",
        "EV_Sales",
        "Month_Name",
        "Year_Month"
    ]
)
```

```python
print(X.shape)
print(y.shape)
```

```python
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
```

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

```python
import joblib
```

```python
joblib.dump(X_train, "../models/X_train.pkl")
joblib.dump(X_test, "../models/X_test.pkl")

joblib.dump(y_train, "../models/y_train.pkl")
joblib.dump(y_test, "../models/y_test.pkl")
```
