## Step 1: Create a Comparison Table
- Use the metrics you recorded from Linear Regression and Random Forest.

```python
import pandas as pd

comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest"
    ],
    "MAE": [
        306.16,
        121.42
    ],
    "RMSE": [
        459.06,
        173.99
    ],
    "R2 Score": [
        0.9750,
        0.9964
    ],
    "MAPE": [
        0.1137,
        0.0675
    ]
})

comparison
```

Step 2: Save the Table

```python
comparison.to_csv(
    "../data/processed/model_comparison.csv",
    index=False
)
```

## Step 3: RMSE Comparison Chart

```python
import plotly.express as px

fig = px.bar(
    comparison,
    x="Model",
    y="RMSE",
    color="Model",
    text_auto=True,
    title="RMSE Comparison"
)

fig.show()
```

Step 4: MAE Comparison

```python
fig = px.bar(
    comparison,
    x="Model",
    y="MAE",
    color="Model",
    text_auto=True,
    title="MAE Comparison"
)

fig.show()
```

```python
fig = px.bar(
    comparison,
    x="Model",
    y="R2 Score",
    color="Model",
    text_auto=True,
    title="R² Score Comparison"
)

fig.show()
```

```python
best_model = comparison.loc[
    comparison["RMSE"].idxmin()
]

print(best_model)
```
